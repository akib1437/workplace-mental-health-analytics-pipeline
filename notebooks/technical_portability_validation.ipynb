## Before running

Place the original downloaded files in `data/external_raw/` with exactly these names: `demographic.csv`, `phq9.csv`, and `gad7.csv`. Do not open, edit, or re-save the raw CSVs in Excel.

```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd()
if not (project_root / 'data' / 'external_raw').exists():
    project_root = project_root.parent
data_dir = project_root / 'data' / 'external_raw'

print('Project folder:', project_root)
print('Files found:', [file.name for file in data_dir.iterdir()] if data_dir.exists() else 'FOLDER NOT FOUND')

for file_name in ['demographic.csv', 'phq9.csv', 'gad7.csv']:
    path = data_dir / file_name
    print(f'\n{file_name} | exists: {path.exists()}')
    if path.exists():
        df = pd.read_csv(path, nrows=5)
        print('Columns:', df.columns.tolist())
        display(df)
```
